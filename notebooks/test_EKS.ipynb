````julia
using EnsembleKalmanProcesses
using EnsembleKalmanProcesses.ParameterDistributions
using LinearAlgebra
using Statistics
using Random
using Plots

Random.seed!(123)  # For reproducibility

# ============================================
# TOY MODEL: Simple Energy Balance
# ============================================
# Model: Temperature evolves as T(t) = T_eq * (1 - exp(-t/τ)) + noise
# Where:
#   - T_eq = equilibrium temperature (depends on climate sensitivity)
#   - τ = time constant (depends on ocean heat uptake)
#
# Parameters to calibrate:
#   θ[1] = climate_sensitivity (affects T_eq)
#   θ[2] = time_constant (affects τ)

function simple_climate_model(θ; n_years=100)
    """
    Simple toy climate model
    θ[1]: climate_sensitivity (true value will be 3.0)
    θ[2]: time_constant (true value will be 20.0)
    """
    climate_sensitivity = θ[1]
    time_constant = θ[2]
    
    # Equilibrium temperature depends on sensitivity
    T_eq = 15.0 + 2.0 * (climate_sensitivity - 3.0)  # Base temp 15°C
    
    # Time evolution
    times = 1:n_years
    temperatures = T_eq .* (1 .- exp.(-times ./ time_constant))
    
    # Add some internal variability
    temperatures .+= 0.1 * randn(n_years)
    
    return temperatures
end

# ============================================
# FORWARD MAP G: Model + Summary Statistics
# ============================================
function G(θ)
    """
    Runs model and computes summary statistics
    Returns: [mean_temp, final_temp, trend, variability]
    """
    temps = simple_climate_model(θ)
    
    n = length(temps)
    times = 1:n
    
    # Compute summary statistics
    mean_temp = mean(temps)
    final_temp = temps[end]
    
    # Linear trend (simple regression)
    trend = (n * sum(times .* temps) - sum(times) * sum(temps)) / 
            (n * sum(times.^2) - sum(times)^2)
    
    variability = std(temps)
    
    return [mean_temp, final_temp, trend, variability]
end

# ============================================
# CREATE "TRUTH" - Our synthetic observations
# ============================================
θ_true = [3.0, 20.0]  # True climate sensitivity and time constant

println("True parameters: ", θ_true)
println("Generating synthetic observations...")

# Generate "observed" data
temps_true = simple_climate_model(θ_true)
y = G(θ_true)  # Our "observations"

println("True summary statistics (observations):")
println("  Mean temperature: ", y[1])
println("  Final temperature: ", y[2])
println("  Trend: ", y[3])
println("  Variability: ", y[4])

# Add observation noise
obs_noise_std = [0.2, 0.3, 0.01, 0.05]  # Uncertainty in each statistic
y_noisy = y .+ obs_noise_std .* randn(length(y))
obs_noise_cov = Diagonal(obs_noise_std.^2)

println("\nNoisy observations:")
println("  y = ", y_noisy)

# ============================================
# SET UP PRIOR
# ============================================
# We'll pretend we don't know the true values
prior = combine_distributions([
    constrained_gaussian("climate_sensitivity", 4.0, 1.5, 1.0, 6.0),  # Wrong initial guess!
    constrained_gaussian("time_constant", 30.0, 10.0, 5.0, 50.0)      # Wrong initial guess!
])

println("\nPrior distributions:")
println("  Climate sensitivity: mean=4.0, std=1.5, bounds=[1.0, 6.0]")
println("  Time constant: mean=30.0, std=10.0, bounds=[5.0, 50.0]")

# ============================================
# RUN ENSEMBLE KALMAN SAMPLER
# ============================================
println("\n" * "="^50)
println("Starting EKS calibration...")
println("="^50)

N_ensemble = 100
N_iterations = 100

# Create initial ensemble
initial_ensemble = construct_initial_ensemble(prior, N_ensemble)

# Initialize EKS
eks_process = Sampler(prior)
eksobj = EnsembleKalmanProcess(
    initial_ensemble, 
    y_noisy, 
    obs_noise_cov, 
    eks_process,
    verbose=true
)

# Store results for plotting
param_history = zeros(2, N_iterations + 1, N_ensemble)
param_history[:, 1, :] = get_ϕ_final(prior, eksobj)

# Main calibration loop
for i in 1:N_iterations
    # Get current parameters (in constrained/physical space)
    params_i = get_ϕ_final(prior, eksobj)
    
    # Evaluate forward map for each ensemble member
    G_ensemble = zeros(length(y_noisy), N_ensemble)
    for j in 1:N_ensemble
        G_ensemble[:, j] = G(params_i[:, j])
    end
    
    # Update ensemble
    update_ensemble!(eksobj, G_ensemble)
    
    # Store for plotting
    param_history[:, i+1, :] = get_ϕ_final(prior, eksobj)
    
    # Print progress
    current_mean = get_ϕ_mean_final(prior, eksobj)
    println("Iteration $i: params = ", round.(current_mean, digits=3))
end

# ============================================
# EXTRACT RESULTS
# ============================================
println("\n" * "="^50)
println("RESULTS")
println("="^50)

# Final parameter estimates
θ_optimal = get_ϕ_mean_final(prior, eksobj)
final_ensemble = get_ϕ_final(prior, eksobj)
θ_std = std(final_ensemble, dims=2)

println("True parameters:      ", round.(θ_true, digits=3))
println("Estimated parameters: ", round.(θ_optimal, digits=3))
println("Parameter std:        ", round.(vec(θ_std), digits=3))
println("\nErrors:")
println("  Climate sensitivity: ", round(θ_optimal[1] - θ_true[1], digits=3))
println("  Time constant:       ", round(θ_optimal[2] - θ_true[2], digits=3))

# Compare forward model outputs
y_estimated = G(θ_optimal)
println("\nForward model comparison:")
println("  Observed:  ", round.(y_noisy, digits=3))
println("  Estimated: ", round.(y_estimated, digits=3))
println("  True:      ", round.(y, digits=3))

# ============================================
# VISUALIZATION
# ============================================
println("\nCreating visualizations...")

# Plot 1: Parameter evolution
p1 = plot(layout=(2,1), size=(800, 600), legend=:right)

for j in 1:N_ensemble
    plot!(p1[1], 0:N_iterations, param_history[1, :, j], 
          alpha=0.3, color=:blue, label=(j==1 ? "Ensemble" : ""))
end
hline!(p1[1], [θ_true[1]], color=:red, linewidth=3, label="True", linestyle=:dash)
plot!(p1[1], 0:N_iterations, vec(mean(param_history[1, :, :], dims=2)), 
      color=:black, linewidth=3, label="Mean")
xlabel!(p1[1], "Iteration")
ylabel!(p1[1], "Climate Sensitivity")
title!(p1[1], "Parameter Evolution: Climate Sensitivity")

for j in 1:N_ensemble
    plot!(p1[2], 0:N_iterations, param_history[2, :, j], 
          alpha=0.3, color=:blue, label=(j==1 ? "Ensemble" : ""))
end
hline!(p1[2], [θ_true[2]], color=:red, linewidth=3, label="True", linestyle=:dash)
plot!(p1[2], 0:N_iterations, vec(mean(param_history[2, :, :], dims=2)), 
      color=:black, linewidth=3, label="Mean")
xlabel!(p1[2], "Iteration")
ylabel!(p1[2], "Time Constant")
title!(p1[2], "Parameter Evolution: Time Constant")

display(p1)

# Plot 2: Final parameter distributions
p2 = plot(layout=(1,2), size=(1000, 400))

histogram!(p2[1], final_ensemble[1, :], bins=20, alpha=0.6, 
           label="Posterior", normalize=:pdf)
vline!(p2[1], [θ_true[1]], color=:red, linewidth=3, 
       label="True", linestyle=:dash)
vline!(p2[1], [θ_optimal[1]], color=:black, linewidth=3, 
       label="Estimated")
xlabel!(p2[1], "Climate Sensitivity")
ylabel!(p2[1], "Density")
title!(p2[1], "Posterior Distribution")

histogram!(p2[2], final_ensemble[2, :], bins=20, alpha=0.6, 
           label="Posterior", normalize=:pdf)
vline!(p2[2], [θ_true[2]], color=:red, linewidth=3, 
       label="True", linestyle=:dash)
vline!(p2[2], [θ_optimal[2]], color=:black, linewidth=3, 
       label="Estimated")
xlabel!(p2[2], "Time Constant")
ylabel!(p2[2], "Density")
title!(p2[2], "Posterior Distribution")

display(p2)

# Plot 3: Model fit - compare temperature time series
p3 = plot(size=(800, 500))

times = 1:100
temps_estimated = simple_climate_model(θ_optimal)
temps_true_clean = simple_climate_model(θ_true, n_years=100)

# Plot several ensemble members
for j in 1:5
    temps_ens = simple_climate_model(final_ensemble[:, j])
    plot!(p3, times, temps_ens, alpha=0.3, color=:blue, 
          label=(j==1 ? "Ensemble members" : ""))
end

plot!(p3, times, temps_true_clean, color=:red, linewidth=3, 
      label="True model", linestyle=:dash)
plot!(p3, times, temps_estimated, color=:black, linewidth=3, 
      label="Estimated (mean)")
xlabel!(p3, "Year")
ylabel!(p3, "Temperature (°C)")
title!(p3, "Model Fit: Temperature Time Series")

display(p3)

println("\n" * "="^50)
println("Calibration complete!")
println("="^50)
````

````julia
using EnsembleKalmanProcesses
using EnsembleKalmanProcesses.ParameterDistributions
using LinearAlgebra
using Statistics
using Random
using Plots
using JLD2
using Dates

Random.seed!(123)

# ============================================
# CONFIGURATION
# ============================================
const OUTPUT_DIR = "eks_results"
const CHECKPOINT_DIR = joinpath(OUTPUT_DIR, "checkpoints")
const FIGURES_DIR = joinpath(OUTPUT_DIR, "figures")

mkpath(OUTPUT_DIR)
mkpath(CHECKPOINT_DIR)
mkpath(FIGURES_DIR)

println("Output directory: ", OUTPUT_DIR)
println("Checkpoint directory: ", CHECKPOINT_DIR)
println("Figures directory: ", FIGURES_DIR)

# ============================================
# TOY MODEL
# ============================================
function simple_climate_model(θ; n_years=100)
    climate_sensitivity = θ[1]
    time_constant = θ[2]
    
    T_eq = 15.0 + 2.0 * (climate_sensitivity - 3.0)
    times = 1:n_years
    temperatures = T_eq .* (1 .- exp.(-times ./ time_constant))
    temperatures .+= 0.1 * randn(n_years)
    
    return temperatures
end

function G(θ)
    temps = simple_climate_model(θ)
    
    n = length(temps)
    times = 1:n
    
    mean_temp = mean(temps)
    final_temp = temps[end]
    trend = (n * sum(times .* temps) - sum(times) * sum(temps)) / 
            (n * sum(times.^2) - sum(times)^2)
    variability = std(temps)
    
    return [mean_temp, final_temp, trend, variability]
end

# ============================================
# CHECKPOINTING FUNCTIONS
# ============================================

function save_checkpoint(iteration, eksobj, prior, param_history, 
                        y_noisy, obs_noise_cov, metadata)
    checkpoint_file = joinpath(CHECKPOINT_DIR, "checkpoint_iter_$(iteration).jld2")
    
    checkpoint_data = Dict(
        "iteration" => iteration,
        "eksobj" => eksobj,
        "prior" => prior,
        "param_history" => param_history,
        "y_noisy" => y_noisy,
        "obs_noise_cov" => obs_noise_cov,
        "metadata" => metadata,
        "timestamp" => now()
    )
    
    @save checkpoint_file checkpoint_data
    println("  ✓ Checkpoint saved: ", checkpoint_file)
    
    return checkpoint_file
end

function load_checkpoint()
    checkpoint_files = filter(f -> startswith(f, "checkpoint_iter_"), 
                             readdir(CHECKPOINT_DIR))
    
    if isempty(checkpoint_files)
        println("No checkpoints found.")
        return nothing
    end
    
    iterations = [parse(Int, match(r"checkpoint_iter_(\d+)", f).captures[1]) 
                 for f in checkpoint_files]
    latest_iter = maximum(iterations)
    latest_file = joinpath(CHECKPOINT_DIR, "checkpoint_iter_$(latest_iter).jld2")
    
    println("Loading checkpoint from iteration $latest_iter...")
    @load latest_file checkpoint_data
    
    return checkpoint_data
end

function save_iteration_results(iteration, params_current, G_ensemble, 
                               param_mean, param_std)
    results_file = joinpath(OUTPUT_DIR, "iteration_results.jld2")
    
    if isfile(results_file)
        @load results_file all_results
    else
        all_results = Dict()
    end
    
    all_results[iteration] = Dict(
        "params_ensemble" => params_current,
        "G_ensemble" => G_ensemble,
        "param_mean" => param_mean,
        "param_std" => param_std,
        "timestamp" => now()
    )
    
    @save results_file all_results
end

function save_final_results(θ_true, θ_optimal, θ_std, final_ensemble,
                           y_noisy, y_estimated, metadata)
    final_file = joinpath(OUTPUT_DIR, "final_results.jld2")
    
    final_data = Dict(
        "θ_true" => θ_true,
        "θ_optimal" => θ_optimal,
        "θ_std" => θ_std,
        "final_ensemble" => final_ensemble,
        "y_noisy" => y_noisy,
        "y_estimated" => y_estimated,
        "metadata" => metadata,
        "timestamp" => now()
    )
    
    @save final_file final_data
    println("Final results saved: ", final_file)
    
    summary_file = joinpath(OUTPUT_DIR, "summary.txt")
    open(summary_file, "w") do io
        println(io, "="^60)
        println(io, "EKS CALIBRATION SUMMARY")
        println(io, "="^60)
        println(io, "Completed: ", now())
        println(io, "")
        println(io, "True parameters:      ", θ_true)
        println(io, "Estimated parameters: ", θ_optimal)
        println(io, "Parameter std:        ", θ_std)
        println(io, "")
        println(io, "Errors:")
        for (i, (true_val, est_val, std_val)) in enumerate(zip(θ_true, θ_optimal, θ_std))
            err = est_val - true_val
            n_sigma = abs(err) / std_val
            println(io, "  Parameter $i: error = $err ($(round(n_sigma, digits=2))σ)")
        end
        println(io, "")
        println(io, "Forward model comparison:")
        println(io, "  Observed:  ", y_noisy)
        println(io, "  Estimated: ", y_estimated)
        println(io, "")
        println(io, "Metadata:")
        for (key, val) in metadata
            println(io, "  $key: $val")
        end
    end
    println("Summary saved: ", summary_file)
end

# ============================================
# MAIN CALIBRATION
# ============================================

function run_calibration(;restart=false, checkpoint_frequency=5, N_iterations=20)
    
    metadata = Dict(
        "start_time" => now(),
        "N_iterations" => N_iterations,
        "checkpoint_frequency" => checkpoint_frequency
    )
    
    if restart
        checkpoint_data = load_checkpoint()
        
        if checkpoint_data === nothing
            println("No checkpoint found. Starting fresh.")
            restart = false
        else
            eksobj = checkpoint_data["eksobj"]
            prior = checkpoint_data["prior"]
            param_history = checkpoint_data["param_history"]
            y_noisy = checkpoint_data["y_noisy"]
            obs_noise_cov = checkpoint_data["obs_noise_cov"]
            start_iteration = checkpoint_data["iteration"] + 1
            metadata = merge(checkpoint_data["metadata"], 
                           Dict("restarted_at" => now()))
            
            println("Restarting from iteration $start_iteration")
        end
    end
    
    if !restart
        println("\n" * "="^60)
        println("STARTING NEW CALIBRATION")
        println("="^60)
        
        θ_true = [3.0, 20.0]
        y = G(θ_true)
        
        obs_noise_std = [0.2, 0.3, 0.01, 0.05]
        y_noisy = y .+ obs_noise_std .* randn(length(y))
        obs_noise_cov = Diagonal(obs_noise_std.^2)
        
        println("True parameters: ", θ_true)
        println("Observations: ", y_noisy)
        
        prior = combine_distributions([
            constrained_gaussian("climate_sensitivity", 4.0, 1.5, 1.0, 6.0),
            constrained_gaussian("time_constant", 30.0, 10.0, 5.0, 50.0)
        ])
        
        N_ensemble = 50
        initial_ensemble = construct_initial_ensemble(prior, N_ensemble)
        
        eks_process = Sampler(prior)
        eksobj = EnsembleKalmanProcess(
            initial_ensemble, 
            y_noisy, 
            obs_noise_cov, 
            eks_process,
            verbose=true
        )
        
        param_history = zeros(2, N_iterations + 1, N_ensemble)
        param_history[:, 1, :] = get_ϕ_final(prior, eksobj)
        
        start_iteration = 1
        
        metadata["θ_true"] = θ_true
        metadata["N_ensemble"] = N_ensemble
        metadata["obs_noise_std"] = obs_noise_std
        
        save_checkpoint(0, eksobj, prior, param_history, 
                       y_noisy, obs_noise_cov, metadata)
    end
    
    println("\n" * "="^60)
    println("RUNNING ITERATIONS $(start_iteration) to $(N_iterations)")
    println("="^60)
    
    for i in start_iteration:N_iterations
        iter_start_time = now()
        println("\n--- Iteration $i/$N_iterations ---")
        
        params_i = get_ϕ_final(prior, eksobj)
        
        println("  Evaluating forward model for $(size(params_i, 2)) ensemble members...")
        N_ensemble = size(params_i, 2)
        G_ensemble = zeros(length(y_noisy), N_ensemble)
        
        for j in 1:N_ensemble
            G_ensemble[:, j] = G(params_i[:, j])
            
            if j % 10 == 0
                println("    Completed $j/$N_ensemble members")
            end
        end
        
        update_ensemble!(eksobj, G_ensemble)
        
        param_history[:, i+1, :] = get_ϕ_final(prior, eksobj)
        
        current_mean = get_ϕ_mean_final(prior, eksobj)
        current_ensemble = get_ϕ_final(prior, eksobj)
        current_std = std(current_ensemble, dims=2)
        
        println("  Current parameters: ", round.(current_mean, digits=3))
        println("  Current std:        ", round.(vec(current_std), digits=3))
        
        iter_time = (now() - iter_start_time).value / 1000
        println("  Iteration time: $(round(iter_time, digits=1)) seconds")
        
        save_iteration_results(i, params_i, G_ensemble, current_mean, current_std)
        
        if i % checkpoint_frequency == 0 || i == N_iterations
            save_checkpoint(i, eksobj, prior, param_history, 
                          y_noisy, obs_noise_cov, metadata)
        end
    end
    
    println("\n" * "="^60)
    println("CALIBRATION COMPLETE - EXTRACTING RESULTS")
    println("="^60)
    
    θ_optimal = get_ϕ_mean_final(prior, eksobj)
    final_ensemble = get_ϕ_final(prior, eksobj)
    θ_std = std(final_ensemble, dims=2)
    
    θ_true = metadata["θ_true"]
    y_estimated = G(θ_optimal)
    
    println("\nTrue parameters:      ", round.(θ_true, digits=3))
    println("Estimated parameters: ", round.(θ_optimal, digits=3))
    println("Parameter std:        ", round.(vec(θ_std), digits=3))
    println("\nErrors:")
    println("  Climate sensitivity: ", round(θ_optimal[1] - θ_true[1], digits=3))
    println("  Time constant:       ", round(θ_optimal[2] - θ_true[2], digits=3))
    
    save_final_results(θ_true, θ_optimal, vec(θ_std), final_ensemble,
                      y_noisy, y_estimated, metadata)
    
    println("\nGenerating plots...")
    create_all_plots(param_history, final_ensemble, θ_true, θ_optimal, 
                    vec(θ_std), y_noisy, N_iterations, N_ensemble)
    
    metadata["end_time"] = now()
    total_time = (metadata["end_time"] - metadata["start_time"]).value / 1000 / 60
    println("\nTotal calibration time: $(round(total_time, digits=2)) minutes")
    
    return eksobj, param_history, metadata
end

# ============================================
# PLOTTING FUNCTIONS
# ============================================

function create_all_plots(param_history, final_ensemble, θ_true, θ_optimal, 
                         θ_std, y_noisy, N_iterations, N_ensemble)
    
    p1 = plot(layout=(2,1), size=(800, 600), legend=:right)
    
    param1_mean = vec(mean(param_history[1, :, :], dims=2))
    param1_std = vec(std(param_history[1, :, :], dims=2))
    
    plot!(p1[1], 0:N_iterations, param1_mean .+ param1_std, 
          fillrange=param1_mean .- param1_std, 
          fillalpha=0.3, fillcolor=:blue, label="±1 std", linewidth=0)
    
    for j in 1:min(10, N_ensemble)
        plot!(p1[1], 0:N_iterations, param_history[1, :, j], 
              alpha=0.2, color=:blue, label="")
    end
    
    hline!(p1[1], [θ_true[1]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    plot!(p1[1], 0:N_iterations, param1_mean, 
          color=:black, linewidth=3, label="Mean")
    xlabel!(p1[1], "Iteration")
    ylabel!(p1[1], "Climate Sensitivity")
    title!(p1[1], "Parameter Evolution: Climate Sensitivity")
    
    param2_mean = vec(mean(param_history[2, :, :], dims=2))
    param2_std = vec(std(param_history[2, :, :], dims=2))
    
    plot!(p1[2], 0:N_iterations, param2_mean .+ param2_std, 
          fillrange=param2_mean .- param2_std, 
          fillalpha=0.3, fillcolor=:blue, label="±1 std", linewidth=0)
    
    for j in 1:min(10, N_ensemble)
        plot!(p1[2], 0:N_iterations, param_history[2, :, j], 
              alpha=0.2, color=:blue, label="")
    end
    
    hline!(p1[2], [θ_true[2]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    plot!(p1[2], 0:N_iterations, param2_mean, 
          color=:black, linewidth=3, label="Mean")
    xlabel!(p1[2], "Iteration")
    ylabel!(p1[2], "Time Constant")
    title!(p1[2], "Parameter Evolution: Time Constant")
    
    savefig(p1, joinpath(FIGURES_DIR, "parameter_evolution.png"))
    println("  ✓ Saved: parameter_evolution.png")
    
    p2 = plot(layout=(1,2), size=(1000, 400))
    
    histogram!(p2[1], final_ensemble[1, :], bins=20, alpha=0.6, 
               label="Posterior", normalize=:pdf, color=:blue)
    vline!(p2[1], [θ_true[1]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    vline!(p2[1], [θ_optimal[1]], color=:black, linewidth=3, label="Mean")
    vline!(p2[1], [θ_optimal[1] - θ_std[1], θ_optimal[1] + θ_std[1]], 
           color=:gray, linewidth=2, linestyle=:dot, label="±1 std")
    annotate!(p2[1], θ_optimal[1], maximum(ylims(p2[1])) * 0.9,
              text("$(round(θ_optimal[1], digits=2)) ± $(round(θ_std[1], digits=3))", 
              :center, 10))
    xlabel!(p2[1], "Climate Sensitivity")
    ylabel!(p2[1], "Density")
    title!(p2[1], "Posterior: Climate Sensitivity")
    
    histogram!(p2[2], final_ensemble[2, :], bins=20, alpha=0.6, 
               label="Posterior", normalize=:pdf, color=:blue)
    vline!(p2[2], [θ_true[2]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    vline!(p2[2], [θ_optimal[2]], color=:black, linewidth=3, label="Mean")
    vline!(p2[2], [θ_optimal[2] - θ_std[2], θ_optimal[2] + θ_std[2]], 
           color=:gray, linewidth=2, linestyle=:dot, label="±1 std")
    annotate!(p2[2], θ_optimal[2], maximum(ylims(p2[2])) * 0.9,
              text("$(round(θ_optimal[2], digits=2)) ± $(round(θ_std[2], digits=2))", 
              :center, 10))
    xlabel!(p2[2], "Time Constant")
    ylabel!(p2[2], "Density")
    title!(p2[2], "Posterior: Time Constant")
    
    savefig(p2, joinpath(FIGURES_DIR, "posterior_distributions.png"))
    println("  ✓ Saved: posterior_distributions.png")
    
    p3 = plot(size=(800, 500))
    
    times = 1:100
    temps_ensemble = zeros(100, N_ensemble)
    for j in 1:N_ensemble
        temps_ensemble[:, j] = simple_climate_model(final_ensemble[:, j])
    end
    
    temps_mean = vec(mean(temps_ensemble, dims=2))
    temps_std = vec(std(temps_ensemble, dims=2))
    
    plot!(p3, times, temps_mean .+ temps_std, 
          fillrange=temps_mean .- temps_std, 
          fillalpha=0.3, fillcolor=:blue, 
          label="±1 std", linewidth=0)
    
    plot!(p3, times, temps_mean .+ 2*temps_std, 
          fillrange=temps_mean .- 2*temps_std, 
          fillalpha=0.15, fillcolor=:blue, 
          label="±2 std", linewidth=0)
    
    temps_true = simple_climate_model(θ_true, n_years=100)
    plot!(p3, times, temps_true, color=:red, linewidth=3, 
          label="True", linestyle=:dash)
    plot!(p3, times, temps_mean, color=:black, linewidth=3, 
          label="Estimated")
    
    xlabel!(p3, "Year")
    ylabel!(p3, "Temperature (°C)")
    title!(p3, "Model Fit: Temperature Time Series")
    
    savefig(p3, joinpath(FIGURES_DIR, "model_fit.png"))
    println("  ✓ Saved: model_fit.png")
    
    p4 = scatter(final_ensemble[1, :], final_ensemble[2, :], 
                 alpha=0.4, label="Posterior samples",
                 xlabel="Climate Sensitivity", ylabel="Time Constant",
                 title="Joint Posterior Distribution",
                 size=(600, 500), color=:blue, markersize=4)
    
    scatter!(p4, [θ_true[1]], [θ_true[2]], 
             color=:red, markersize=10, markershape=:star5, label="True")
    scatter!(p4, [θ_optimal[1]], [θ_optimal[2]], 
             color=:black, markersize=8, markershape=:cross, label="Estimated")
    
    savefig(p4, joinpath(FIGURES_DIR, "joint_posterior.png"))
    println("  ✓ Saved: joint_posterior.png")
    
    p5 = plot(layout=(2,1), size=(800, 600))
    
    std_history_param1 = vec(std(param_history[1, :, :], dims=2))
    std_history_param2 = vec(std(param_history[2, :, :], dims=2))
    
    plot!(p5[1], 0:N_iterations, std_history_param1,
          color=:blue, linewidth=3, label="Parameter std",
          xlabel="Iteration", ylabel="Standard Deviation",
          title="Uncertainty Evolution: Climate Sensitivity")
    
    plot!(p5[2], 0:N_iterations, std_history_param2,
          color=:blue, linewidth=3, label="Parameter std",
          xlabel="Iteration", ylabel="Standard Deviation",
          title="Uncertainty Evolution: Time Constant")
    
    savefig(p5, joinpath(FIGURES_DIR, "uncertainty_evolution.png"))
    println("  ✓ Saved: uncertainty_evolution.png")
end

# ============================================
# RUN IT!
# ============================================

eksobj, param_history, metadata = run_calibration(
    restart=false,
    checkpoint_frequency=5,
    N_iterations=20
)

println("\n" * "="^60)
println("ALL DONE! Results saved in: $OUTPUT_DIR")
println("="^60)
````

````julia
using JLD2
using Plots
using Printf
using Statistics
using LinearAlgebra

# ============================================
# LOAD RESULTS AND CREATE ANIMATIONS
# ============================================

function load_final_results(output_dir="eks_results")
    """Load the final results from a completed calibration"""
    final_file = joinpath(output_dir, "final_results.jld2")
    
    if !isfile(final_file)
        error("No final results found in $output_dir")
    end
    
    @load final_file final_data
    return final_data
end

function load_checkpoint_for_animation(checkpoint_file)
    """Load a specific checkpoint"""
    @load checkpoint_file checkpoint_data
    return checkpoint_data
end

function create_convergence_animation(output_dir="eks_results"; fps=2)
    """
    Create an animation showing parameter convergence over iterations
    """
    
    println("Loading results...")
    final_data = load_final_results(output_dir)
    
    checkpoint_files = filter(f -> startswith(f, "checkpoint_iter_"), 
                             readdir(joinpath(output_dir, "checkpoints")))
    iterations = [parse(Int, match(r"checkpoint_iter_(\d+)", f).captures[1]) 
                 for f in checkpoint_files]
    latest_iter = maximum(iterations)
    latest_file = joinpath(output_dir, "checkpoints", "checkpoint_iter_$(latest_iter).jld2")
    
    checkpoint_data = load_checkpoint_for_animation(latest_file)
    param_history = checkpoint_data["param_history"]
    
    θ_true = final_data["θ_true"]
    N_iterations = size(param_history, 2) - 1
    N_ensemble = size(param_history, 3)
    
    println("Creating animation...")
    println("  Number of iterations: $N_iterations")
    println("  Ensemble size: $N_ensemble")
    
    anim = @animate for iter in 0:N_iterations
        params_at_iter = param_history[:, iter+1, :]
        
        param_mean = vec(mean(params_at_iter, dims=2))
        param_std = vec(std(params_at_iter, dims=2))
        
        p = scatter(params_at_iter[1, :], params_at_iter[2, :],
                   alpha=0.5,
                   color=:blue,
                   markersize=6,
                   label="Ensemble members",
                   xlabel="Climate Sensitivity",
                   ylabel="Time Constant",
                   title=@sprintf("Iteration %d/%d", iter, N_iterations),
                   xlims=(1.5, 6.0),
                   ylims=(5.0, 45.0),
                   size=(700, 600),
                   legend=:topright)
        
        scatter!(p, [θ_true[1]], [θ_true[2]], 
                color=:red, 
                markersize=12, 
                markershape=:star5,
                label="True value")
        
        scatter!(p, [param_mean[1]], [param_mean[2]], 
                color=:black, 
                markersize=10, 
                markershape=:cross,
                linewidth=3,
                label="Ensemble mean")
        
        if iter > 0
            cov_matrix = cov(params_at_iter')
            eigen_result = eigen(cov_matrix)
            angle = atan(eigen_result.vectors[2,1], eigen_result.vectors[1,1])
            width = 2 * sqrt(eigen_result.values[1])
            height = 2 * sqrt(eigen_result.values[2])
            
            θ_ellipse = range(0, 2π, length=100)
            ellipse_x = param_mean[1] .+ width * cos.(θ_ellipse) * cos(angle) - 
                                       height * sin.(θ_ellipse) * sin(angle)
            ellipse_y = param_mean[2] .+ width * cos.(θ_ellipse) * sin(angle) + 
                                       height * sin.(θ_ellipse) * cos(angle)
            
            plot!(p, ellipse_x, ellipse_y, 
                 color=:blue, 
                 linewidth=2, 
                 linestyle=:dash,
                 label="~1σ contour")
        end
        
        annotate!(p, 4.5, 42, 
                 text(@sprintf("Mean: [%.2f, %.2f]\nStd: [%.2f, %.2f]", 
                      param_mean[1], param_mean[2], 
                      param_std[1], param_std[2]), 
                      :right, 9))
    end
    
    # Display in notebook instead of saving
    display(gif(anim, fps=fps))
    println("✓ Animation displayed!")
    
    return anim
end

function create_time_series_animation(output_dir="eks_results"; fps=2)
    """
    Create animation showing parameter evolution over time (1D time series)
    """
    
    println("Loading results...")
    final_data = load_final_results(output_dir)
    
    checkpoint_files = filter(f -> startswith(f, "checkpoint_iter_"), 
                             readdir(joinpath(output_dir, "checkpoints")))
    iterations = [parse(Int, match(r"checkpoint_iter_(\d+)", f).captures[1]) 
                 for f in checkpoint_files]
    latest_iter = maximum(iterations)
    latest_file = joinpath(output_dir, "checkpoints", "checkpoint_iter_$(latest_iter).jld2")
    
    checkpoint_data = load_checkpoint_for_animation(latest_file)
    param_history = checkpoint_data["param_history"]
    
    θ_true = final_data["θ_true"]
    N_iterations = size(param_history, 2) - 1
    N_ensemble = size(param_history, 3)
    
    println("Creating time series animation...")
    
    anim = @animate for current_iter in 0:N_iterations
        p = plot(layout=(2,1), size=(800, 700))
        
        # Parameter 1
        param1_mean = vec(mean(param_history[1, 1:current_iter+1, :], dims=2))
        param1_std = vec(std(param_history[1, 1:current_iter+1, :], dims=2))
        
        if current_iter > 0
            plot!(p[1], 0:current_iter, param1_mean .+ param1_std, 
                  fillrange=param1_mean .- param1_std, 
                  fillalpha=0.3, fillcolor=:blue, 
                  label="±1 std", linewidth=0)
        end
        
        for j in 1:min(10, N_ensemble)
            plot!(p[1], 0:current_iter, param_history[1, 1:current_iter+1, j], 
                  alpha=0.2, color=:blue, label="")
        end
        
        plot!(p[1], 0:current_iter, param1_mean, 
              color=:black, linewidth=3, label="Mean")
        
        hline!(p[1], [θ_true[1]], color=:red, linewidth=3, 
               label="True", linestyle=:dash)
        
        xlabel!(p[1], "Iteration")
        ylabel!(p[1], "Climate Sensitivity")
        title!(p[1], @sprintf("Iteration %d/%d", current_iter, N_iterations))
        xlims!(p[1], (0, N_iterations))
        ylims!(p[1], (1.5, 6.0))
        
        # Parameter 2
        param2_mean = vec(mean(param_history[2, 1:current_iter+1, :], dims=2))
        param2_std = vec(std(param_history[2, 1:current_iter+1, :], dims=2))
        
        if current_iter > 0
            plot!(p[2], 0:current_iter, param2_mean .+ param2_std, 
                  fillrange=param2_mean .- param2_std, 
                  fillalpha=0.3, fillcolor=:blue, 
                  label="±1 std", linewidth=0)
        end
        
        for j in 1:min(10, N_ensemble)
            plot!(p[2], 0:current_iter, param_history[2, 1:current_iter+1, j], 
                  alpha=0.2, color=:blue, label="")
        end
        
        plot!(p[2], 0:current_iter, param2_mean, 
              color=:black, linewidth=3, label="Mean")
        
        hline!(p[2], [θ_true[2]], color=:red, linewidth=3, 
               label="True", linestyle=:dash)
        
        xlabel!(p[2], "Iteration")
        ylabel!(p[2], "Time Constant")
        xlims!(p[2], (0, N_iterations))
        ylims!(p[2], (5.0, 45.0))
    end
    
    # Display in notebook instead of saving
    display(gif(anim, fps=fps))
    println("✓ Animation displayed!")
    
    return anim
end

function create_combined_animation(output_dir="eks_results"; fps=2)
    """
    Create a combined animation with both 2D parameter space and 1D time series
    """
    
    println("Loading results...")
    final_data = load_final_results(output_dir)
    
    checkpoint_files = filter(f -> startswith(f, "checkpoint_iter_"), 
                             readdir(joinpath(output_dir, "checkpoints")))
    iterations = [parse(Int, match(r"checkpoint_iter_(\d+)", f).captures[1]) 
                 for f in checkpoint_files]
    latest_iter = maximum(iterations)
    latest_file = joinpath(output_dir, "checkpoints", "checkpoint_iter_$(latest_iter).jld2")
    
    checkpoint_data = load_checkpoint_for_animation(latest_file)
    param_history = checkpoint_data["param_history"]
    
    θ_true = final_data["θ_true"]
    N_iterations = size(param_history, 2) - 1
    N_ensemble = size(param_history, 3)
    
    println("Creating combined animation...")
    
    anim = @animate for current_iter in 0:N_iterations
        p = plot(layout=@layout([a{0.5w} [b; c]]), size=(1200, 800))
        
        # LEFT PANEL: 2D parameter space
        params_at_iter = param_history[:, current_iter+1, :]
        param_mean = vec(mean(params_at_iter, dims=2))
        
        scatter!(p[1], params_at_iter[1, :], params_at_iter[2, :],
                alpha=0.5, color=:blue, markersize=6,
                xlabel="Climate Sensitivity",
                ylabel="Time Constant",
                title=@sprintf("Parameter Space (Iter %d/%d)", current_iter, N_iterations),
                xlims=(1.5, 6.0), ylims=(5.0, 45.0),
                legend=false)
        
        scatter!(p[1], [θ_true[1]], [θ_true[2]], 
                color=:red, markersize=12, markershape=:star5)
        scatter!(p[1], [param_mean[1]], [param_mean[2]], 
                color=:black, markersize=10, markershape=:cross, linewidth=3)
        
        if current_iter > 0
            cov_matrix = cov(params_at_iter')
            eigen_result = eigen(cov_matrix)
            angle = atan(eigen_result.vectors[2,1], eigen_result.vectors[1,1])
            width = 2 * sqrt(eigen_result.values[1])
            height = 2 * sqrt(eigen_result.values[2])
            
            θ_ellipse = range(0, 2π, length=100)
            ellipse_x = param_mean[1] .+ width * cos.(θ_ellipse) * cos(angle) - 
                                       height * sin.(θ_ellipse) * sin(angle)
            ellipse_y = param_mean[2] .+ width * cos.(θ_ellipse) * sin(angle) + 
                                       height * sin.(θ_ellipse) * cos(angle)
            
            plot!(p[1], ellipse_x, ellipse_y, color=:blue, linewidth=2, linestyle=:dash)
        end
        
        # TOP RIGHT: Climate sensitivity evolution
        param1_mean = vec(mean(param_history[1, 1:current_iter+1, :], dims=2))
        param1_std = vec(std(param_history[1, 1:current_iter+1, :], dims=2))
        
        if current_iter > 0
            plot!(p[2], 0:current_iter, param1_mean .+ param1_std, 
                  fillrange=param1_mean .- param1_std, 
                  fillalpha=0.3, fillcolor=:blue, linewidth=0, label="")
        end
        
        for j in 1:min(5, N_ensemble)
            plot!(p[2], 0:current_iter, param_history[1, 1:current_iter+1, j], 
                  alpha=0.15, color=:blue, label="")
        end
        
        plot!(p[2], 0:current_iter, param1_mean, 
              color=:black, linewidth=3, label="")
        hline!(p[2], [θ_true[1]], color=:red, linewidth=3, linestyle=:dash, label="")
        
        ylabel!(p[2], "Climate Sensitivity")
        title!(p[2], "Parameter Evolution")
        xlims!(p[2], (0, N_iterations))
        ylims!(p[2], (1.5, 6.0))
        
        # BOTTOM RIGHT: Time constant evolution
        param2_mean = vec(mean(param_history[2, 1:current_iter+1, :], dims=2))
        param2_std = vec(std(param_history[2, 1:current_iter+1, :], dims=2))
        
        if current_iter > 0
            plot!(p[3], 0:current_iter, param2_mean .+ param2_std, 
                  fillrange=param2_mean .- param2_std, 
                  fillalpha=0.3, fillcolor=:blue, linewidth=0, label="")
        end
        
        for j in 1:min(5, N_ensemble)
            plot!(p[3], 0:current_iter, param_history[2, 1:current_iter+1, j], 
                  alpha=0.15, color=:blue, label="")
        end
        
        plot!(p[3], 0:current_iter, param2_mean, 
              color=:black, linewidth=3, label="")
        hline!(p[3], [θ_true[2]], color=:red, linewidth=3, linestyle=:dash, label="")
        
        xlabel!(p[3], "Iteration")
        ylabel!(p[3], "Time Constant")
        xlims!(p[3], (0, N_iterations))
        ylims!(p[3], (5.0, 45.0))
    end
    
    # Display in notebook instead of saving
    display(gif(anim, fps=fps))
    println("✓ Animation displayed!")
    
    return anim
end

# ============================================
# USAGE
# ============================================

# Display animations directly in notebook
create_convergence_animation("eks_results", fps=3)
create_time_series_animation("eks_results", fps=3)
create_combined_animation("eks_results", fps=3)

println("\n✓ All animations displayed!")
````

# Workflow for HPC with submitting jobs 

````julia
using EnsembleKalmanProcesses
using EnsembleKalmanProcesses.ParameterDistributions
using LinearAlgebra
using Statistics
using JLD2
using Dates

# ============================================
# JOB MANAGEMENT SYSTEM
# ============================================

"""
Structure to track job status
"""
mutable struct JobTracker
    job_id::String
    member_id::Int
    iteration::Int
    status::Symbol  # :submitted, :running, :completed, :failed
    submit_time::DateTime
    completion_time::Union{DateTime, Nothing}
    param_file::String
    output_file::String
end

"""
Write parameters for a single ensemble member
"""
function write_params_for_member(θ, iteration, member_id, work_dir)
    param_file = joinpath(work_dir, "iter_$(iteration)", "member_$(member_id)_params.txt")
    mkpath(dirname(param_file))
    
    open(param_file, "w") do f
        # Adjust format to match your climate model's input
        for (i, val) in enumerate(θ)
            println(f, "param_$i = $val")
        end
    end
    
    return param_file
end

"""
Create SLURM job script for a single ensemble member
"""
function create_job_script(iteration, member_id, param_file, output_dir, work_dir)
    output_file = joinpath(output_dir, "iter_$(iteration)", "member_$(member_id)_output.nc")
    mkpath(dirname(output_file))
    
    script_file = joinpath(work_dir, "iter_$(iteration)", "member_$(member_id)_job.sh")
    
    script_content = """
    #!/bin/bash
    #SBATCH --job-name=eks_i$(iteration)_m$(member_id)
    #SBATCH --time=7-00:00:00
    #SBATCH --ntasks=1
    #SBATCH --cpus-per-task=4
    #SBATCH --mem=16G
    #SBATCH --output=$(work_dir)/iter_$(iteration)/member_$(member_id)_%j.log
    #SBATCH --error=$(work_dir)/iter_$(iteration)/member_$(member_id)_%j.err
    
    # Load modules (adjust for your system)
    # module load netcdf
    # module load your-climate-model
    
    # Run your climate model
    # Replace this with your actual command
    ./your_climate_model \\
        --config $(param_file) \\
        --output $(output_file)
    
    # Check if output was created successfully
    if [ -f $(output_file) ]; then
        echo "SUCCESS" > $(dirname(output_file))/member_$(member_id).status
    else
        echo "FAILED" > $(dirname(output_file))/member_$(member_id).status
        exit 1
    fi
    """
    
    open(script_file, "w") do f
        write(f, script_content)
    end
    
    return script_file, output_file
end

"""
Submit a single job and return job ID
"""
function submit_job(script_file)
    # Submit job and capture job ID
    # Note: --parsable returns just the job ID
    try
        job_id = strip(read(`sbatch --parsable $(script_file)`, String))
        return job_id
    catch e
        @warn "Failed to submit job: $script_file"
        @warn e
        return nothing
    end
end

"""
Check status of a single job
"""
function check_job_status(job_id)
    try
        # Query SLURM for job status
        output = read(`sacct -j $(job_id) -o State -n --parsable2`, String)
        status_line = strip(split(output, "\n")[1])
        
        if occursin("COMPLETED", status_line)
            return :completed
        elseif occursin("RUNNING", status_line)
            return :running
        elseif occursin("PENDING", status_line)
            return :submitted
        elseif occursin("FAILED", status_line) || occursin("CANCELLED", status_line)
            return :failed
        else
            return :unknown
        end
    catch e
        # If sacct fails, job might not be in system yet
        return :unknown
    end
end

"""
Submit all ensemble members for one iteration
"""
function submit_iteration_jobs(params_i, iteration, work_dir, output_dir)
    N_ensemble = size(params_i, 2)
    job_trackers = JobTracker[]
    
    println("  Submitting $N_ensemble jobs for iteration $iteration...")
    
    for j in 1:N_ensemble
        # Write parameters
        param_file = write_params_for_member(params_i[:, j], iteration, j, work_dir)
        
        # Create job script
        script_file, output_file = create_job_script(iteration, j, param_file, output_dir, work_dir)
        
        # Submit job
        job_id = submit_job(script_file)
        
        if job_id !== nothing
            tracker = JobTracker(
                job_id,
                j,
                iteration,
                :submitted,
                now(),
                nothing,
                param_file,
                output_file
            )
            push!(job_trackers, tracker)
            
            if j % 10 == 0 || j == N_ensemble
                println("    Submitted $j/$N_ensemble jobs")
            end
        else
            error("Failed to submit job for member $j")
        end
        
        # Small delay to avoid overwhelming scheduler
        sleep(0.1)
    end
    
    println("  ✓ All $N_ensemble jobs submitted!")
    
    return job_trackers
end

"""
Wait for all jobs in an iteration to complete
"""
function wait_for_iteration_completion(job_trackers; 
                                       check_interval_minutes=30,
                                       max_wait_days=10)
    
    println("\n  Waiting for jobs to complete...")
    println("  (Checking every $check_interval_minutes minutes)")
    
    start_time = now()
    max_wait = Dates.Day(max_wait_days)
    
    completed_count = 0
    failed_count = 0
    
    while true
        # Update status for all jobs
        for tracker in job_trackers
            if tracker.status in [:submitted, :running]
                new_status = check_job_status(tracker.job_id)
                
                if new_status == :completed && tracker.status != :completed
                    tracker.status = :completed
                    tracker.completion_time = now()
                    completed_count += 1
                    println("    ✓ Job $(tracker.member_id) completed ($completed_count/$(length(job_trackers)))")
                    
                elseif new_status == :failed && tracker.status != :failed
                    tracker.status = :failed
                    tracker.completion_time = now()
                    failed_count += 1
                    @warn "Job $(tracker.member_id) failed!"
                    
                elseif new_status == :running && tracker.status == :submitted
                    tracker.status = :running
                    println("    → Job $(tracker.member_id) is now running")
                end
            end
        end
        
        # Check if all done
        n_completed = count(t -> t.status == :completed, job_trackers)
        n_failed = count(t -> t.status == :failed, job_trackers)
        n_pending = count(t -> t.status in [:submitted, :running], job_trackers)
        
        println("    Status: $n_completed completed, $n_failed failed, $n_pending pending")
        
        # All completed
        if n_completed == length(job_trackers)
            println("  ✓ All jobs completed successfully!")
            return :success
        end
        
        # Some failed
        if n_failed > 0 && n_pending == 0
            @warn "Some jobs failed! Completed: $n_completed, Failed: $n_failed"
            return :partial_failure
        end
        
        # Timeout check
        elapsed = now() - start_time
        if elapsed > max_wait
            @warn "Maximum wait time exceeded ($max_wait_days days)"
            return :timeout
        end
        
        # Wait before next check
        sleep(check_interval_minutes * 60)
    end
end

"""
Collect results from completed jobs
"""
function collect_iteration_results(job_trackers, y_obs)
    N_ensemble = length(job_trackers)
    n_outputs = length(y_obs)
    G_ensemble = zeros(n_outputs, N_ensemble)
    
    println("  Collecting results from $N_ensemble jobs...")
    
    for (j, tracker) in enumerate(job_trackers)
        if tracker.status == :completed
            try
                # Read output and compute summary statistics
                G_ensemble[:, j] = compute_summary_statistics(tracker.output_file)
                
                if j % 10 == 0
                    println("    Processed $j/$N_ensemble outputs")
                end
            catch e
                @warn "Failed to process output for member $(tracker.member_id)"
                @warn e
                # Set to NaN - EKS can handle failed members
                G_ensemble[:, j] .= NaN
            end
        else
            @warn "Member $(tracker.member_id) did not complete successfully"
            G_ensemble[:, j] .= NaN
        end
    end
    
    # Check how many failed
    n_failed = sum(isnan.(G_ensemble[1, :]))
    if n_failed > 0
        @warn "$n_failed ensemble members failed"
    end
    
    println("  ✓ Results collected!")
    
    return G_ensemble
end

"""
Compute summary statistics from model output
"""
function compute_summary_statistics(output_file)
    """
    Replace this with your actual implementation
    """
    
    # Example using NCDatasets for NetCDF
    # using NCDatasets
    # 
    # ds = Dataset(output_file, "r")
    # temp = ds["temperature"][:]
    # precip = ds["precipitation"][:]
    # ice = ds["sea_ice_extent"][:]
    # # ... etc
    # close(ds)
    # 
    # # Compute statistics
    # stats = [
    #     mean(temp),
    #     std(temp),
    #     mean(precip),
    #     trend(ice),
    #     # ... etc
    # ]
    
    # Placeholder - replace with actual computation
    if !isfile(output_file)
        error("Output file not found: $output_file")
    end
    
    stats = rand(6)  # REPLACE THIS
    
    return stats
end

"""
Save job tracker information
"""
function save_job_trackers(job_trackers, iteration, output_dir)
    tracker_file = joinpath(output_dir, "job_tracking", "iter_$(iteration)_trackers.jld2")
    mkpath(dirname(tracker_file))
    
    @save tracker_file job_trackers
    
    # Also save human-readable log
    log_file = joinpath(output_dir, "job_tracking", "iter_$(iteration)_log.txt")
    open(log_file, "w") do f
        println(f, "Iteration $iteration Job Tracking")
        println(f, "="^60)
        for tracker in job_trackers
            println(f, "Member $(tracker.member_id):")
            println(f, "  Job ID: $(tracker.job_id)")
            println(f, "  Status: $(tracker.status)")
            println(f, "  Submitted: $(tracker.submit_time)")
            if tracker.completion_time !== nothing
                println(f, "  Completed: $(tracker.completion_time)")
                duration = tracker.completion_time - tracker.submit_time
                println(f, "  Duration: $(duration)")
            end
            println(f, "  Output: $(tracker.output_file)")
            println(f, "")
        end
    end
end

# ============================================
# MAIN CALIBRATION WITH JOB MANAGEMENT
# ============================================

function run_calibration_with_queue(;
    restart=false,
    checkpoint_frequency=1,
    N_iterations=10,
    N_ensemble=25,
    output_dir="eks_climate_calibration",
    work_dir="eks_work",
    check_interval_minutes=30,
    max_wait_days=10)
    
    # Setup directories
    mkpath(output_dir)
    mkpath(work_dir)
    checkpoint_dir = joinpath(output_dir, "checkpoints")
    mkpath(checkpoint_dir)
    
    println("="^60)
    println("EKS CALIBRATION WITH QUEUED JOBS")
    println("="^60)
    println("Output directory: $output_dir")
    println("Work directory: $work_dir")
    println("N_ensemble: $N_ensemble")
    println("N_iterations: $N_iterations")
    println("="^60)
    
    # Setup (prior, observations, etc.)
    prior = combine_distributions([
        constrained_gaussian("param1", 3.0, 1.0, 1.0, 6.0),
        constrained_gaussian("param2", 2.0, 0.5, 0.5, 4.0),
        constrained_gaussian("param3", 1.5, 0.8, 0.0, 3.0),
        constrained_gaussian("param4", 4.0, 1.2, 2.0, 8.0),
        constrained_gaussian("param5", 0.5, 0.2, 0.0, 1.0),
        constrained_gaussian("param6", 10.0, 3.0, 5.0, 20.0)
    ])
    
    # Your observations
    y_obs = [14.8, 1.2, 1000.0, -0.15, 500.0, 280.0]  # Replace with real
    obs_noise_cov = Diagonal([0.2^2, 0.1^2, 100^2, 0.05^2, 50^2, 5^2])
    
    # Initialize EKS
    initial_ensemble = construct_initial_ensemble(prior, N_ensemble)
    eks_process = Sampler(prior)
    eksobj = EnsembleKalmanProcess(
        initial_ensemble,
        y_obs,
        obs_noise_cov,
        eks_process,
        verbose=true
    )
    
    param_history = zeros(6, N_iterations + 1, N_ensemble)
    param_history[:, 1, :] = get_ϕ_final(prior, eksobj)
    
    metadata = Dict(
        "start_time" => now(),
        "N_iterations" => N_iterations,
        "N_ensemble" => N_ensemble,
        "output_dir" => output_dir,
        "work_dir" => work_dir
    )
    
    # Save initial checkpoint
    save_checkpoint(0, eksobj, prior, param_history, 
                   y_obs, obs_noise_cov, metadata, checkpoint_dir)
    
    # Main iteration loop
    for i in 1:N_iterations
        iter_start_time = now()
        
        println("\n" * "="^60)
        println("ITERATION $i/$N_iterations")
        println("="^60)
        
        # Get current parameters
        params_i = get_ϕ_final(prior, eksobj)
        
        # STEP 1: Submit all jobs
        job_trackers = submit_iteration_jobs(params_i, i, work_dir, output_dir)
        
        # Save job tracking info
        save_job_trackers(job_trackers, i, output_dir)
        
        # STEP 2: Wait for completion
        result = wait_for_iteration_completion(job_trackers; 
                                              check_interval_minutes=check_interval_minutes,
                                              max_wait_days=max_wait_days)
        
        if result == :timeout
            error("Iteration $i timed out after $max_wait_days days")
        elseif result == :partial_failure
            @warn "Some jobs failed in iteration $i. Continuing with available results..."
        end
        
        # STEP 3: Collect results
        G_ensemble = collect_iteration_results(job_trackers, y_obs)
        
        # STEP 4: Update EKS
        println("\n  Updating ensemble...")
        update_ensemble!(eksobj, G_ensemble)
        param_history[:, i+1, :] = get_ϕ_final(prior, eksobj)
        
        # Print statistics
        current_mean = get_ϕ_mean_final(prior, eksobj)
        current_std = std(get_ϕ_final(prior, eksobj), dims=2)
        
        println("\n  Parameter estimates:")
        for p in 1:6
            println("    param_$p: $(round(current_mean[p], digits=3)) ± $(round(current_std[p], digits=3))")
        end
        
        iter_duration = now() - iter_start_time
        println("\n  Iteration duration: $(iter_duration)")
        
        # Save results
        save_iteration_results(i, params_i, G_ensemble, 
                             current_mean, current_std, output_dir)
        
        # Checkpoint
        if i % checkpoint_frequency == 0 || i == N_iterations
            save_checkpoint(i, eksobj, prior, param_history,
                          y_obs, obs_noise_cov, metadata, checkpoint_dir)
        end
        
        # Estimate remaining time
        avg_iter_time = (now() - metadata["start_time"]).value / i / 1000 / 60 / 60 / 24  # days
        remaining = (N_iterations - i) * avg_iter_time
        println("\n  Estimated time remaining: $(round(remaining, digits=1)) days")
    end
    
    # Final results
    θ_optimal = get_ϕ_mean_final(prior, eksobj)
    final_ensemble = get_ϕ_final(prior, eksobj)
    θ_std = std(final_ensemble, dims=2)
    
    save_final_results(θ_optimal, vec(θ_std), final_ensemble,
                      y_obs, metadata, output_dir)
    
    println("\n" * "="^60)
    println("CALIBRATION COMPLETE!")
    println("="^60)
    println("\nFinal parameter estimates:")
    for p in 1:6
        println("  param_$p: $(round(θ_optimal[p], digits=3)) ± $(round(θ_std[p], digits=3))")
    end
    
    return eksobj, param_history, metadata
end

# Helper functions (simplified versions - expand as needed)
function save_iteration_results(i, params, G_ensemble, mean_vals, std_vals, output_dir)
    results_file = joinpath(output_dir, "iteration_results.jld2")
    
    if isfile(results_file)
        @load results_file all_results
    else
        all_results = Dict()
    end
    
    all_results[i] = Dict(
        "params" => params,
        "G_ensemble" => G_ensemble,
        "mean" => mean_vals,
        "std" => std_vals,
        "timestamp" => now()
    )
    
    @save results_file all_results
end

function save_checkpoint(i, eksobj, prior, param_history, y_obs, obs_noise_cov, metadata, checkpoint_dir)
    checkpoint_file = joinpath(checkpoint_dir, "checkpoint_iter_$(i).jld2")
    
    checkpoint_data = Dict(
        "iteration" => i,
        "eksobj" => eksobj,
        "prior" => prior,
        "param_history" => param_history,
        "y_obs" => y_obs,
        "obs_noise_cov" => obs_noise_cov,
        "metadata" => metadata,
        "timestamp" => now()
    )
    
    @save checkpoint_file checkpoint_data
    println("  ✓ Checkpoint saved: $checkpoint_file")
end

function save_final_results(θ_optimal, θ_std, final_ensemble, y_obs, metadata, output_dir)
    final_file = joinpath(output_dir, "final_results.jld2")
    
    final_data = Dict(
        "θ_optimal" => θ_optimal,
        "θ_std" => θ_std,
        "final_ensemble" => # Run the full calibration
eksobj, param_history, metadata = run_calibration_with_queue(
    restart=false,
    checkpoint_frequency=1,       # Save after every iteration
    N_iterations=10,
    N_ensemble=25,
    output_dir="my_climate_calibration",
    work_dir="my_climate_work",
    check_interval_minutes=30,    # Check job status every 30 min
    max_wait_days=10              # Timeout after 10 days per iteration
)
```

## Key Features

1. **Asynchronous job submission**: Submits all jobs at once
2. **Status monitoring**: Checks job status periodically
3. **Failure handling**: Tracks which jobs fail, continues with available results
4. **Timeout protection**: Won't wait forever
5. **Full logging**: Saves all job information for debugging
6. **Restartable**: If Julia crashes, you can restart from checkpoint

## Directory Structure
```
my_climate_calibration/
├── checkpoints/
│   ├── checkpoint_iter_0.jld2
│   ├── checkpoint_iter_1.jld2
│   └── ...
├── job_tracking/
│   ├── iter_1_trackers.jld2
│   ├── iter_1_log.txt
│   └── ...
├── iter_1/
│   ├── member_1_output.nc
│   ├── member_2_output.nc
│   └── ...
├── iteration_results.jld2
└── final_results.jld2

my_climate_work/
├── iter_1/
│   ├── member_1_params.txt
│   ├── member_1_job.sh
│   ├── member_1_12345.log
│   └── ...
└── ...final_ensemble,
        "y_obs" => y_obs,
        "metadata" => metadata,
        "timestamp" => now()
    )
    
    @save final_file final_data
    println("Final results saved: $final_file")
end
````

````julia
using JLD2
using Plots
using Statistics
using LinearAlgebra
using Printf

"""
Load all results and create publication-ready plots
"""
function create_all_plots_from_results(output_dir="eks_toy_test")
    
    println("Loading results from $output_dir...")
    
    # Load final results
    @load joinpath(output_dir, "final_results.jld2") final_data
    
    θ_optimal = final_data["θ_optimal"]
    θ_std = final_data["θ_std"]
    final_ensemble = final_data["final_ensemble"]
    y_obs = final_data["y_obs"]
    θ_true = final_data["metadata"]["θ_true"]
    
    # Load checkpoint with complete history
    checkpoint_files = filter(f -> startswith(f, "checkpoint_iter_"), 
                             readdir(joinpath(output_dir, "checkpoints")))
    iterations = [parse(Int, match(r"checkpoint_iter_(\d+)", f).captures[1]) 
                 for f in checkpoint_files]
    latest_iter = maximum(iterations)
    
    @load joinpath(output_dir, "checkpoints", "checkpoint_iter_$(latest_iter).jld2") checkpoint_data
    
    param_history = checkpoint_data["param_history"]
    N_iterations = size(param_history, 2) - 1
    N_ensemble = size(param_history, 3)
    
    println("  Loaded data: $N_iterations iterations, $N_ensemble ensemble members")
    println("  True parameters: $θ_true")
    println("  Estimated parameters: $θ_optimal")
    println("  Parameter uncertainties: $θ_std")
    
    # Create figures directory
    figures_dir = joinpath(output_dir, "figures")
    mkpath(figures_dir)
    
    # ========================================
    # PLOT 1: Parameter Evolution
    # ========================================
    println("\nCreating parameter evolution plot...")
    
    p1 = plot(layout=(2,1), size=(800, 600), legend=:right)
    
    # Climate sensitivity evolution
    param1_mean = vec(mean(param_history[1, :, :], dims=2))
    param1_std = vec(std(param_history[1, :, :], dims=2))
    
    plot!(p1[1], 0:N_iterations, param1_mean .+ param1_std, 
          fillrange=param1_mean .- param1_std, 
          fillalpha=0.3, fillcolor=:blue, label="±1 std", linewidth=0)
    
    for j in 1:min(10, N_ensemble)
        plot!(p1[1], 0:N_iterations, param_history[1, :, j], 
              alpha=0.2, color=:blue, label="")
    end
    
    hline!(p1[1], [θ_true[1]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    plot!(p1[1], 0:N_iterations, param1_mean, 
          color=:black, linewidth=3, label="Mean")
    xlabel!(p1[1], "Iteration")
    ylabel!(p1[1], "Climate Sensitivity")
    title!(p1[1], "Parameter Evolution: Climate Sensitivity")
    
    # Time constant evolution
    param2_mean = vec(mean(param_history[2, :, :], dims=2))
    param2_std = vec(std(param_history[2, :, :], dims=2))
    
    plot!(p1[2], 0:N_iterations, param2_mean .+ param2_std, 
          fillrange=param2_mean .- param2_std, 
          fillalpha=0.3, fillcolor=:blue, label="±1 std", linewidth=0)
    
    for j in 1:min(10, N_ensemble)
        plot!(p1[2], 0:N_iterations, param_history[2, :, j], 
              alpha=0.2, color=:blue, label="")
    end
    
    hline!(p1[2], [θ_true[2]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    plot!(p1[2], 0:N_iterations, param2_mean, 
          color=:black, linewidth=3, label="Mean")
    xlabel!(p1[2], "Iteration")
    ylabel!(p1[2], "Time Constant")
    title!(p1[2], "Parameter Evolution: Time Constant")
    
    savefig(p1, joinpath(figures_dir, "parameter_evolution.png"))
    println("  ✓ Saved: parameter_evolution.png")
    
    # ========================================
    # PLOT 2: Posterior Distributions
    # ========================================
    println("Creating posterior distribution plot...")
    
    p2 = plot(layout=(1,2), size=(1000, 400))
    
    # Climate sensitivity posterior
    histogram!(p2[1], final_ensemble[1, :], bins=15, alpha=0.6, 
               label="Posterior", normalize=:pdf, color=:blue)
    vline!(p2[1], [θ_true[1]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    vline!(p2[1], [θ_optimal[1]], color=:black, linewidth=3, label="Mean")
    vline!(p2[1], [θ_optimal[1] - θ_std[1], θ_optimal[1] + θ_std[1]], 
           color=:gray, linewidth=2, linestyle=:dot, label="±1 std")
    annotate!(p2[1], θ_optimal[1], maximum(ylims(p2[1])) * 0.9,
              text(@sprintf("%.2f ± %.3f", θ_optimal[1], θ_std[1]), :center, 10))
    xlabel!(p2[1], "Climate Sensitivity")
    ylabel!(p2[1], "Density")
    title!(p2[1], "Posterior: Climate Sensitivity")
    
    # Time constant posterior
    histogram!(p2[2], final_ensemble[2, :], bins=15, alpha=0.6, 
               label="Posterior", normalize=:pdf, color=:blue)
    vline!(p2[2], [θ_true[2]], color=:red, linewidth=3, 
           label="True", linestyle=:dash)
    vline!(p2[2], [θ_optimal[2]], color=:black, linewidth=3, label="Mean")
    vline!(p2[2], [θ_optimal[2] - θ_std[2], θ_optimal[2] + θ_std[2]], 
           color=:gray, linewidth=2, linestyle=:dot, label="±1 std")
    annotate!(p2[2], θ_optimal[2], maximum(ylims(p2[2])) * 0.9,
              text(@sprintf("%.2f ± %.2f", θ_optimal[2], θ_std[2]), :center, 10))
    xlabel!(p2[2], "Time Constant")
    ylabel!(p2[2], "Density")
    title!(p2[2], "Posterior: Time Constant")
    
    savefig(p2, joinpath(figures_dir, "posterior_distributions.png"))
    println("  ✓ Saved: posterior_distributions.png")
    
    # ========================================
    # PLOT 3: Model Fit
    # ========================================
    println("Creating model fit plot...")
    
    # Need to reconstruct model outputs
    # Simple climate model function (same as in test script)
    function simple_climate_model(θ; n_years=100)
        climate_sensitivity = θ[1]
        time_constant = θ[2]
        
        T_eq = 15.0 + 2.0 * (climate_sensitivity - 3.0)
        times = 1:n_years
        temperatures = T_eq .* (1 .- exp.(-times ./ time_constant))
        
        # Don't add noise for visualization
        return temperatures
    end
    
    p3 = plot(size=(800, 500))
    
    times = 1:100
    
    # Generate predictions for all ensemble members
    temps_ensemble = zeros(100, N_ensemble)
    for j in 1:N_ensemble
        temps_ensemble[:, j] = simple_climate_model(final_ensemble[:, j])
    end
    
    temps_mean = vec(mean(temps_ensemble, dims=2))
    temps_std = vec(std(temps_ensemble, dims=2))
    
    # Plot uncertainty bands
    plot!(p3, times, temps_mean .+ temps_std, 
          fillrange=temps_mean .- temps_std, 
          fillalpha=0.3, fillcolor=:blue, 
          label="±1 std", linewidth=0)
    
    plot!(p3, times, temps_mean .+ 2*temps_std, 
          fillrange=temps_mean .- 2*temps_std, 
          fillalpha=0.15, fillcolor=:blue, 
          label="±2 std", linewidth=0)
    
    # True model
    temps_true = simple_climate_model(θ_true, n_years=100)
    plot!(p3, times, temps_true, color=:red, linewidth=3, 
          label="True", linestyle=:dash)
    
    # Estimated model
    plot!(p3, times, temps_mean, color=:black, linewidth=3, 
          label="Estimated")
    
    xlabel!(p3, "Year")
    ylabel!(p3, "Temperature (°C)")
    title!(p3, "Model Fit: Temperature Time Series")
    
    savefig(p3, joinpath(figures_dir, "model_fit.png"))
    println("  ✓ Saved: model_fit.png")
    
    # ========================================
    # PLOT 4: Joint Posterior
    # ========================================
    println("Creating joint posterior plot...")
    
    p4 = scatter(final_ensemble[1, :], final_ensemble[2, :], 
                 alpha=0.4, label="Posterior samples",
                 xlabel="Climate Sensitivity", ylabel="Time Constant",
                 title="Joint Posterior Distribution",
                 size=(600, 500), color=:blue, markersize=6)
    
    # Add 1-sigma contour ellipse
    cov_matrix = cov(final_ensemble')
    eigen_result = eigen(cov_matrix)
    angle = atan(eigen_result.vectors[2,1], eigen_result.vectors[1,1])
    width = 2 * sqrt(eigen_result.values[1])
    height = 2 * sqrt(eigen_result.values[2])
    
    θ_ellipse = range(0, 2π, length=100)
    ellipse_x = θ_optimal[1] .+ width * cos.(θ_ellipse) * cos(angle) - 
                               height * sin.(θ_ellipse) * sin(angle)
    ellipse_y = θ_optimal[2] .+ width * cos.(θ_ellipse) * sin(angle) + 
                               height * sin.(θ_ellipse) * cos(angle)
    
    plot!(p4, ellipse_x, ellipse_y, color=:blue, linewidth=2, 
          linestyle=:dash, label="~1σ contour")
    
    # Mark true and estimated
    scatter!(p4, [θ_true[1]], [θ_true[2]], 
             color=:red, markersize=12, markershape=:star5, label="True")
    scatter!(p4, [θ_optimal[1]], [θ_optimal[2]], 
             color=:black, markersize=10, markershape=:cross, label="Estimated")
    
    savefig(p4, joinpath(figures_dir, "joint_posterior.png"))
    println("  ✓ Saved: joint_posterior.png")
    
    # ========================================
    # PLOT 5: Uncertainty Evolution
    # ========================================
    println("Creating uncertainty evolution plot...")
    
    p5 = plot(layout=(2,1), size=(800, 600))
    
    std_history_param1 = vec(std(param_history[1, :, :], dims=2))
    std_history_param2 = vec(std(param_history[2, :, :], dims=2))
    
    plot!(p5[1], 0:N_iterations, std_history_param1,
          color=:blue, linewidth=3, label="Parameter std",
          xlabel="Iteration", ylabel="Standard Deviation",
          title="Uncertainty Evolution: Climate Sensitivity",
          marker=:circle, markersize=4)
    
    plot!(p5[2], 0:N_iterations, std_history_param2,
          color=:blue, linewidth=3, label="Parameter std",
          xlabel="Iteration", ylabel="Standard Deviation",
          title="Uncertainty Evolution: Time Constant",
          marker=:circle, markersize=4)
    
    savefig(p5, joinpath(figures_dir, "uncertainty_evolution.png"))
    println("  ✓ Saved: uncertainty_evolution.png")
    
    # ========================================
    # PLOT 6: Error vs Iteration
    # ========================================
    println("Creating error evolution plot...")
    
    p6 = plot(size=(800, 500))
    
    # Compute error (distance from true parameters) at each iteration
    errors = zeros(N_iterations + 1)
    for i in 1:(N_iterations + 1)
        mean_at_i = vec(mean(param_history[:, i, :], dims=2))
        errors[i] = sqrt(sum((mean_at_i - θ_true).^2))
    end
    
    plot!(p6, 0:N_iterations, errors,
          color=:blue, linewidth=3, marker=:circle,
          label="Parameter Error",
          xlabel="Iteration",
          ylabel="L2 Distance from True Parameters",
          title="Convergence to True Parameters",
          yscale=:log10)
    
    savefig(p6, joinpath(figures_dir, "error_convergence.png"))
    println("  ✓ Saved: error_convergence.png")
    
    # ========================================
    # Summary Statistics
    # ========================================
    println("\n" * "="^60)
    println("SUMMARY STATISTICS")
    println("="^60)
    
    summary_file = joinpath(figures_dir, "summary.txt")
    open(summary_file, "w") do f
        println(f, "="^60)
        println(f, "EKS CALIBRATION RESULTS SUMMARY")
        println(f, "="^60)
        println(f, "")
        println(f, "True parameters:")
        println(f, "  Climate sensitivity: $(θ_true[1])")
        println(f, "  Time constant:       $(θ_true[2])")
        println(f, "")
        println(f, "Estimated parameters:")
        println(f, "  Climate sensitivity: $(round(θ_optimal[1], digits=3)) ± $(round(θ_std[1], digits=3))")
        println(f, "  Time constant:       $(round(θ_optimal[2], digits=3)) ± $(round(θ_std[2], digits=3))")
        println(f, "")
        println(f, "Errors:")
        println(f, "  Climate sensitivity: $(round(θ_optimal[1] - θ_true[1], digits=3)) ($(round((θ_optimal[1] - θ_true[1])/θ_std[1], digits=2))σ)")
        println(f, "  Time constant:       $(round(θ_optimal[2] - θ_true[2], digits=3)) ($(round((θ_optimal[2] - θ_true[2])/θ_std[2], digits=2))σ)")
        println(f, "")
        println(f, "Calibration info:")
        println(f, "  Number of iterations: $N_iterations")
        println(f, "  Ensemble size: $N_ensemble")
        println(f, "  Final error: $(round(errors[end], digits=4))")
        println(f, "  Error reduction: $(round((errors[1] - errors[end])/errors[1] * 100, digits=1))%")
    end
    
    println("\nFinal Results:")
    println("  True:      $θ_true")
    println("  Estimated: ", round.(θ_optimal, digits=3))
    println("  Std:       ", round.(θ_std, digits=3))
    println("  Error:     ", round.(θ_optimal - θ_true, digits=3))
    
    println("\n" * "="^60)
    println("All plots saved to: $figures_dir")
    println("="^60)
    println("\nCreated plots:")
    println("  1. parameter_evolution.png")
    println("  2. posterior_distributions.png")
    println("  3. model_fit.png")
    println("  4. joint_posterior.png")
    println("  5. uncertainty_evolution.png")
    println("  6. error_convergence.png")
    println("  7. summary.txt")
end

# Run it!
create_all_plots_from_results("eks_toy_test")
````
